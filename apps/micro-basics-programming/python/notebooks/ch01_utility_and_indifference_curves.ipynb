## 試してみよう

- `a = 2` にすると、(20, 40) の効用は 16000、MRS は 4.0 になります。MRS は、
  **a ÷ b と、y ÷ x** で決まります
- `a = 0.5`、`b = 0.5` にすると、効用は約 28.28 ですが、MRS は 2.0 のまま変わりません
  （a ÷ b が、1 のままだから）
- `x = 40`、`y = 20` にすると、MRS は 0.5 に下がります。右へ進むほど MRS が小さい
  （逓減する）ので、曲線は**原点に凸**です
- `a = 2` で `x_at = 10` にすると、曲線の y は 8.0 です（a = 1 のときは 80）
